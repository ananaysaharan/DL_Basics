```python
import tiktoken

# Load the clk100k base tokenizer
enc = tiktoken.get_encoding("cl100k_base")

print("Tokenizer loaded successfully.")
```

```python
text_to_encode = "hello world"

# Encode the text
tokens = enc.encode(text_to_encode)

# Decode tokens to show original words (if applicable, tiktoken breaks into subword units)
# For simple words like 'hello' and 'world', it might return the words themselves or parts.
decoded_tokens = [enc.decode_single_token_bytes(t).decode('utf-8') for t in tokens]

# Get the number of tokens
num_tokens = len(tokens)

print(f"Original text: '{text_to_encode}'")
print(f"Tokens (IDs): {tokens}")
print(f"Decoded tokens (words/subwords): {decoded_tokens}")
print(f"Number of tokens: {num_tokens}")
```

### Exploring Tokenization with Different Examples

Let's try encoding various types of text to observe how the `cl100k_base` tokenizer handles them. This will include sentences with punctuation, numbers, and different casing.

```python
def analyze_tokenization(text):
    tokens = enc.encode(text)
    decoded_tokens = [enc.decode_single_token_bytes(t).decode('utf-8', errors='ignore') for t in tokens]
    num_tokens = len(tokens)

    print(f"\n--- Analyzing: '{text}' ---")
    print(f"Tokens (IDs): {tokens}")
    print(f"Decoded tokens (words/subwords): {decoded_tokens}")
    print(f"Number of tokens: {num_tokens}")

# Example 1: Sentence with punctuation and numbers
analyze_tokenization("GPT-4 is a powerful language model with 175 billion parameters.")

# Example 2: Text with mixed casing and special characters
analyze_tokenization("It's a beautiful day! How are you?")

# Example 3: A URL or code snippet (often tokenized differently)
analyze_tokenization("https://www.example.com/page?id=123")

# Example 4: A more complex phrase with potential subword tokenization
analyze_tokenization("antidisestablishmentarianism")
```

### Comparing Token Efficiency Across Content Types

Now, let's add examples for JSON and code snippets, and then we'll summarize the tokenization behavior and efficiency across all the different content types we've seen.

```python
# Example 5: A JSON snippet
json_snippet = '''{
    "name": "John Doe",
    "age": 30,
    "isStudent": false,
    "courses": ["Math", "Science"]
}'''
analyze_tokenization(json_snippet)

# Example 6: A Python code snippet
python_code = """def greet(name):
    print(f"Hello, {name}!")
greet("Alice")"""
analyze_tokenization(python_code)
```

### Token Efficiency Summary

Now, let's summarize the tokenization behavior and efficiency across all the different content types by calculating 'characters per token'.

```python
def calculate_efficiency(text, name):
    tokens = enc.encode(text)
    num_tokens = len(tokens)
    num_characters = len(text)

    if num_tokens > 0:
        chars_per_token = num_characters / num_tokens
    else:
        chars_per_token = 0

    print(f"\n--- Efficiency for: '{name}' ---")
    print(f"Text Length (characters): {num_characters}")
    print(f"Number of Tokens: {num_tokens}")
    print(f"Characters per Token: {chars_per_token:.2f}")

# Examples from previous analysis
calculate_efficiency(text_to_encode, "'hello world'")
calculate_efficiency("GPT-4 is a powerful language model with 175 billion parameters.", "GPT-4 sentence")
calculate_efficiency("It's a beautiful day! How are you?", "Mixed casing sentence")
calculate_efficiency("https://www.example.com/page?id=123", "URL example")
calculate_efficiency("antidisestablishmentarianism", "Long word")
calculate_efficiency(json_snippet, "JSON snippet")
calculate_efficiency(python_code, "Python code snippet")
```

### Demonstrating the Token Boundary Problem

The 'token boundary problem' refers to the challenge LLMs face when performing tasks that require character-level reasoning, especially when the original word is split into multiple subword tokens. Since the model operates on tokens, not individual characters, it can become difficult to perform operations like counting characters, reversing words, or identifying specific character patterns within a word if that word is fragmented.

```python
# Let's take a word that is likely to be broken into subword tokens
word_to_analyze = "strawberry"

print(f"\n--- Analyzing the word: '{word_to_analyze}' ---")

tokens_word = enc.encode(word_to_analyze)
decoded_tokens_word = [enc.decode_single_token_bytes(t).decode('utf-8', errors='ignore') for t in tokens_word]

print(f"Original word: '{word_to_analyze}'")
print(f"Tokens (IDs): {tokens_word}")
print(f"Decoded tokens: {decoded_tokens_word}")
print(f"Number of tokens: {len(tokens_word)}")
print(f"Number of characters in original word: {len(word_to_analyze)}")

# Now, consider a task like 'count the number of 'r's in the word'
# If the model only 'sees' the decoded tokens, it's not straightforward to count character occurrences
# across token boundaries or within subword tokens without reassembling the word.
```

As you can see, the word 'strawberry' is broken into multiple tokens (`['straw', 'berry']`). If a model were asked to count the number of 'r's in 'strawberry' and it only processed these tokens, it would need to infer or reconstruct the original word to get the correct count (3 'r's). Directly counting 'r's in `['straw', 'berry']` would be ambiguous or incorrect without additional processing logic, highlighting the token boundary problem.
